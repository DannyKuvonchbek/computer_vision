```python
from google.colab.patches import cv2_imshow
import cv2

image = cv2.imread("/content/cat image.jpg")
cv2_imshow(image)
```

```python
print ("SHAPE:", image.shape)
# 350 => height
# 700 => width
# 3 => color channels (BGR) Blue,Green,Red
print("")
```

```python
(b, g, r) = image[100,50]
print("Blue:",b,"Green:",g, "Red:",r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized  = cv2.resize(gray, (300,200))
cv2_imshow(resized)
```

```python
cropped = gray [50:200,300:500]
cv2_imshow(cropped)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped.png", flipped)
```

```python
flipped_cat = cv2.imread("/content/flipped.png")
cv2_imshow(flipped_cat)
```
